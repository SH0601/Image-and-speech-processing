```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img = cv2.imread(r"C:\Users\super\Downloads\png-clipart-gold-colored-j-capital-letter-j-miscellaneous-alphabet.png", 0)
kernel = np.ones((5,5), np.uint8)

erosion = cv2.erode(img, kernel, iterations=1)
dilation = cv2.dilate(img, kernel, iterations=1)

plt.subplot(131)
plt.imshow(img)
plt.title("Original Image")

plt.subplot(132)
plt.imshow(erosion)
plt.title("Eroded Image")

plt.subplot(133)
plt.imshow(dilation)
plt.title("Dilated Image")

plt.show()
```
