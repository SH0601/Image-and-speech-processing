```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img = cv2.imread(r"C:\Users\super\Downloads\images (1).jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.subplot(121)
plt.imshow(img, 'gray')

sift = cv2.SIFT_create()
kp = sift.detect(img, None)

img = cv2.drawKeypoints(img, kp, None)

plt.subplot(122)
plt.imshow(img, 'gray')

plt.show()
```
