```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\super\Downloads\striking-close-up-baboon-its-distinctive-red-rear-untamed-african-captivating-photograph-captures-essence-349053009.webp", 0)

filter = np.array([[-1, 0, 1],
                   [-2, 0, 2],
                   [-1, 0, 1]])

sobel_verti_image = cv2.filter2D(img, -1, filter)
sobel_hori_image = cv2.filter2D(img, -1, np.flip(filter.T, axis=0))

plt.subplot(121)
plt.imshow(sobel_verti_image, cmap='gray')
plt.title("Vertical Edge")

plt.subplot(122)
plt.imshow(sobel_hori_image, cmap='gray')
plt.title("Horizontal Edge")

plt.show()
```
