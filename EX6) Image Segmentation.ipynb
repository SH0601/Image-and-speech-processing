```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img = cv2.imread(r"C:\Users\super\Downloads\images.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print(img.shape)

plt.subplot(121)
plt.imshow(img)
plt.title('Original Image')

pix = img.reshape((-1, 3))
pix = np.float32(pix)

cri = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 100, 0.85)

k = 3

ret, lab, cen = cv2.kmeans(pix, k, None, cri, 10, cv2.KMEANS_RANDOM_CENTERS)

cen = np.uint8(cen)

seg = cen[lab.flatten()]

seg_img = seg.reshape(img.shape)

print(seg_img.shape)

plt.subplot(122)
plt.imshow(seg_img)
plt.title('Clustered Image')

plt.show()
```
