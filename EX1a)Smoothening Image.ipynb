```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img=cv2.imread(r"C:\Users\super\OneDrive\Desktop\Fav actor\silverstone-125930595-16x9_0.webp")
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

blur=cv2.blur(img,(5,5))

plt.subplot(121)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(122)
plt.imshow(blur)
plt.title("Blured")
plt.axis('off')

plt.show()

```
