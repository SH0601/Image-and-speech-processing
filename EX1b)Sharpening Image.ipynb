```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

img=cv2.imread(r"C:\Users\super\OneDrive\Desktop\Fav actor\silverstone-125930595-16x9_0.webp")
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

filter=np.array([[-1,-1,-1],
                [-1,9,-1],
                [-1,-1,-1]])
sharpen_img=cv2.filter2D(img,-1,filter)
plt.subplot(121)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(122)
plt.imshow(sharpen_img)
plt.title("Sharpening")
plt.axis('off')

plt.show()

```
