```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\super\Downloads\images (1).jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

R, G, B = cv2.split(img)

ycrcb = cv2.cvtColor(img, cv2.COLOR_RGB2YCrCb)
Y, Cr, Cb = cv2.split(ycrcb)

hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
H, S, V = cv2.split(hsv)

I = (R.astype(float) + G + B) / 3
I = I.astype(np.uint8)

plt.figure(figsize=(12,6))

plt.subplot(2,5,1)
plt.imshow(R, 'gray')
plt.title('Red')
plt.axis('off')

plt.subplot(2,5,2)
plt.imshow(G, 'gray')
plt.title('Green')
plt.axis('off')

plt.subplot(2,5,3)
plt.imshow(B, 'gray')
plt.title('Blue')
plt.axis('off')

plt.subplot(2,5,4)
plt.imshow(Y, 'gray')
plt.title('Luma (Y)')
plt.axis('off')

plt.subplot(2,5,5)
plt.imshow(V, 'gray')
plt.title('Value (V)')
plt.axis('off')

plt.subplot(2,5,6)
plt.imshow(Cr, 'gray')
plt.title('Chroma Red (Cr)')
plt.axis('off')

plt.subplot(2,5,7)
plt.imshow(Cb, 'gray')
plt.title('Chroma Blue (Cb)')
plt.axis('off')

plt.subplot(2,5,8)
plt.imshow(H, 'gray')
plt.title('Hue')
plt.axis('off')

plt.subplot(2,5,9)
plt.imshow(S, 'gray')
plt.title('Saturation')
plt.axis('off')

plt.subplot(2,5,10)
plt.imshow(I, 'gray')
plt.title('Intensity')
plt.axis('off')

plt.tight_layout()
plt.show()
```
